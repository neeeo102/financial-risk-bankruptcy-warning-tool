```python
# ---------------------------------------------------------------------------------------------------
# 16. Outlier Analysis
# ---------------------------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("16. OUTLIER ANALYSIS")
print("=" * 80)

def calculate_outlier_ratio(series):
    series = series.dropna()
    if len(series) == 0:
        return np.nan
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outlier_count = ((series < lower) | (series > upper)).sum()
    return outlier_count / len(series)

numeric_cols = [col for col in df.select_dtypes(include=[np.number]).columns if col != target_col]

outlier_summary = []
for col in numeric_cols:
    ratio = calculate_outlier_ratio(df[col])
    outlier_summary.append([col, ratio])

outlier_df = pd.DataFrame(outlier_summary, columns=["Feature", "Outlier_Ratio"])
outlier_df = outlier_df.sort_values(by="Outlier_Ratio", ascending=False).reset_index(drop=True)

print("Top 20 features with the highest outlier ratios:")
display(outlier_df.head(20))
outlier_df.to_csv("outputs/tables/outlier_summary.csv", index=False)

plt.figure(figsize=(10, 6))
sns.barplot(data=outlier_df.head(15), x="Outlier_Ratio", y="Feature", palette="magma")
plt.title("Top 15 Features with Highest Outlier Ratio")
plt.xlabel("Outlier Ratio")
plt.ylabel("Feature")
plt.tight_layout()
plt.savefig("outputs/figures/top15_outlier_ratio.png", dpi=300)
plt.show()

print("""
Interpretation:
Outlier analysis shows that some financial indicators contain a relatively
high proportion of extreme values. This is common in financial ratio data
and supports the use of robust analytical methods.
""")
```

```python
# ---------------------------------------------------------------------------------------------------
# 17. Statistical Significance Test
# ---------------------------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("17. STATISTICAL SIGNIFICANCE TEST")
print("=" * 80)

group_0 = df[df[target_col] == 0]
group_1 = df[df[target_col] == 1]

test_results = []

for col in key_features:
    if col in ["Liability_Assets_Flag", "Net_Income_Flag"]:
        continue
    try:
        x0 = group_0[col].dropna()
        x1 = group_1[col].dropna()

        if len(x0) > 0 and len(x1) > 0:
            stat, p_value = mannwhitneyu(x0, x1, alternative="two-sided")
            test_results.append({
                "Feature": col,
                "P_Value": p_value,
                "Significant_at_0_05": p_value < 0.05
            })
    except Exception as e:
        print(f"Skipped {col}: {e}")

test_df = pd.DataFrame(test_results).sort_values(by="P_Value").reset_index(drop=True)

print("Mann–Whitney U test results:")
display(test_df)
test_df.to_csv("outputs/tables/mannwhitney_test_results.csv", index=False)

significant_features = test_df[test_df["Significant_at_0_05"] == True]
print(f"Number of statistically significant features at the 5% level: {len(significant_features)}")

plt.figure(figsize=(10, 6))
plot_df = test_df.copy()
plot_df["Negative_Log10_P"] = -np.log10(plot_df["P_Value"])

sns.barplot(
    data=plot_df.sort_values("Negative_Log10_P", ascending=False),
    x="Negative_Log10_P",
    y="Feature",
    palette="viridis"
)

plt.axvline(-np.log10(0.05), color="red", linestyle="--", label="p = 0.05")
plt.title("Statistical Significance of Key Features")
plt.xlabel("-log10(p-value)")
plt.ylabel("Feature")
plt.legend()
plt.tight_layout()
plt.savefig("outputs/figures/statistical_significance_features.png", dpi=300)
plt.show()

print("""
Interpretation:
The Mann–Whitney U test indicates whether the distributions of key financial
indicators differ significantly between bankrupt and non-bankrupt firms.
Statistically significant results provide stronger support for the relevance
of these variables in bankruptcy risk analysis.
""")
```
